# 3. Merge resume text with submission metadata

Normalizes the emoji-laden headers of the form export and joins it with the extracted resume text. The merged file is the raw submission batch that notebook 04 ingests.

```python
import pandas as pd
import re

def normalize_header(header):
    """Strip emoji / symbols from a column name, trim, snake_case and lowercase it."""
    cleaned = re.sub(r'[^\w\s]', '', header)
    cleaned = cleaned.strip().replace(' ', '_')
    return cleaned.lower()

# 1. read the extracted resume text and the original submission export
df_extracted = pd.read_csv('Extracted_Resumes.csv')
df_submission = pd.read_csv('Submission.csv')

# 2. normalize headers (e.g. '🧑‍🎓 Name' -> 'name', '* Position' -> 'position')
df_extracted.columns = [normalize_header(col) for col in df_extracted.columns]
df_submission.columns = [normalize_header(col) for col in df_submission.columns]

# 3. outer join on name so no submission is dropped
df_merged = pd.merge(df_extracted, df_submission, on='name', how='outer')

# 4. save
output_filename = 'Merged_Resumes_Normalized.csv'
df_merged.to_csv(output_filename, index=False, encoding='utf-8-sig')
print(f"Saved {output_filename}: {len(df_merged)} rows")
```
